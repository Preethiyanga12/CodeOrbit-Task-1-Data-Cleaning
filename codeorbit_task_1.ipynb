```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Hours_Studied': [
        1, 2, 3, 4, 5, 6, 7, 8, 2, 3,
        4, 5, 6, 7, 8, 9, 10, 4, 6, 7
    ],

    'Marks': [
        35, 40, 45, 50, 55, 60, 65, 80, 42, 47,
        52, 58, 62, 70, 85, 90, 95, 53, 63, 72
    ],

    'Department': [
        'AI', 'CSE', 'IT', 'AI', 'CSE',
        'IT', 'AI', 'CSE', 'IT', 'AI',
        'CSE', 'IT', 'AI', 'CSE', 'IT',
        'AI', 'CSE', 'IT', 'AI', 'CSE'
    ],

    'Attendance': [
        80, 85, 90, 75, 88, 92, 85, 95, 78, 82,
        89, 91, 87, 93, 96, 98, 97, 84, 90, 94
    ]
}

df = pd.DataFrame(data)

# Add some data problems for cleaning
df.loc[2, 'Marks'] = np.nan
df.loc[5, 'Attendance'] = np.nan
df.loc[7, 'Hours_Studied'] = np.nan
df.loc[10, 'Hours_Studied'] = 'four'

# Add duplicate row
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

# Save dataset as CSV
df.to_csv('student_data.csv', index=False)

print(df)
```

```python
df = pd.read_csv('student_data.csv')

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df['Marks'] = df['Marks'].fillna(df['Marks'].median())

df['Attendance'] = df['Attendance'].fillna(
    df['Attendance'].mean()
)

df['Hours_Studied'] = pd.to_numeric(
    df['Hours_Studied'], errors='coerce'
)

df['Hours_Studied'] = df['Hours_Studied'].fillna(
    df['Hours_Studied'].median()
)
```

```python
df = df.drop_duplicates()

print("Duplicates removed")
```

```python
print(df.dtypes)
```

```python
df['Hours_Studied'] = df['Hours_Studied'].astype(float)

df['Marks'] = df['Marks'].astype(float)

df['Attendance'] = df['Attendance'].astype(float)

print(df.dtypes)
```

```python
print("Mean:")
print(df.mean(numeric_only=True))

print("Median:")
print(df.median(numeric_only=True))

print("Value Counts:")
print(df['Department'].value_counts())

print("Statistical Summary:")
print(df.describe())
```
